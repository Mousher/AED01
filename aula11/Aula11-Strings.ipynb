2. Faça um algoritmo que solicite uma data no formato de uma string – dd/mm/aaaa.  Mostre essa data no formato AAAAMMDD

```python
data = input("digite uma data no formato dd/mm/aaaa: ")
dia = data[0:2]
mes = data[3:5]
ano = data[6:]
nova_data = ano + mes + dia
print(nova_data)
```

3. Elabore um algoritmo para determinar quantas vogais existem dentro de uma determinada frase (que deve ser recebida do usuário).

```python
frase = input("Digite uma frase: ")
vogal = 0
for letra in frase.lower():
    if letra == 'a' or letra == 'e' or letra == 'i' or letra == 'o' or letra == 'u':
        vogal += 1
        
print("A quantidade de vogais é: ",vogal)
```

4. Faça um algoritmo para determinar quantas palavras existem em uma determinada frase Obs: tanto a palavra, quanto a frase, devem ser informadas pelo usuário.

```python
palavra = input("Digite a palavra desejada: ")
texto = input("Digite o texto que precisa buscar as palavras: ")
num = texto.count(palavra)
print("A palavra ", palavra,"se repete ", num, "vezes no texto.")

```

5. Faça um algoritmo para determinar se um determinada frase, digitada pelo usuário, é um palíndroma. Palíndromo: lido da direita para a esquerda, ou vice versa, representam a mesma coisa.
Ex: AMA

```python
frase = input("Digite uma palavra: ")

frase_limpa = ''
for letra in frase:
    if letra != ' ':
        frase_limpa += letra.lower()

if frase_limpa == frase_limpa[::-1]:
    print("A palavra é palíndromo!")
else:
    print("A palavra não é palíndromo!")
```

6. Faça um algoritmo para ler nove caracteres numéricos em uma string. Mostre o conteúdo dessa string colando pontos e virgula, respectivamente nas posições inteiras e decimais.

Exemplo: 

Digitado> 987654321

Mostrado> 9.876.543,21


```python
numeros = input("Entre com 9 números: ")
n_final = ' '
for n in range(len(numeros)):
    if n == 0 or n == 3:
        n_final = n_final + numeros[n] + '.'
    elif n == 6:
        n_final = n_final + numeros[n] + ','
    else:
        n_final = n_final + numeros[n]
print(n_final)
```
